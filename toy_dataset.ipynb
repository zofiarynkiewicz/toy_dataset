```python
import os
from toy_dataset_repo.helpers import *
import json
import random
```

```python
def generate_dataset(n_samples= 1000, n_digits= 4, seed= 42):

    format_functions = {
        "standard": format_standard,
        "spaced": format_spaced,
        "10e": format_10e,
        "underscore": format_underscore,
        "leading_zeros": format_leading_zeros,
        "comma": comma_separator
    }

    pairs = set()
    random.seed(seed)
    dataset_list = []

    while len(pairs) < n_samples:
        a, b = generate_two_numbers(n_digits=n_digits)
        pair = tuple(sorted((a, b)))

        if pair in pairs:
            continue
        pairs.add(pair)

        for pt in ["standard", "spaced", "10e", "underscore", "comma", "leading_zeros"]:
            entry = format_functions[pt](a, b)
            entry["a"] = a
            entry["b"] = b
            dataset_list.append(entry)

    return dataset_list
```

```python
def save_dataset(dataset, path_json):
    dir_name = os.path.dirname(path_json)
    if dir_name:
        os.makedirs(dir_name, exist_ok=True)
    with open(path_json, "w") as f:
        json.dump(dataset, f, indent=2)
```

```python
ds = generate_dataset(n_samples=1000, n_digits=4, seed=42)
save_dataset(ds, path_json="dataset/toy-dataset.json")
```
